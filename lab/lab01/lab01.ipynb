## 1.6. Ядро (Kernel)
Ядро — це програма, яка виконує код у вашому блокноті та виводить результати. У верхньому правому куті вікна ви можете побачити коло, якє вказує на стан вашого ядра. Якщо кружечок порожній (⚪), ядро ​​неактивне та готове до виконання коду. Якщо кружечок заповнений (⚫), ядро ​​зайняте виконанням коду. 

Біля кожної комірки коду ви побачите текст «[ ]». Перед тим, як запустити клітинку, ви побачите`[ ]`. Коли клітинка запрацює, ви побачите `[*]`. Якщо ви бачите зірочку (\*) поруч із коміркою, яка не зникає, імовірно, код у комірці виконується надто довго, і якщо, щось пішло не так можна перервати роботу ядра (про це йдеться нижче). Коли комірка закінчить виконання коду, ви побачите число в дужках, наприклад: `[1]`. Номер відповідає порядку, в якому ви запускаєте комірки; тому перша комірка, яку ви запустите, покаже 1, коли вона буде виконана, друга покаже 2 і так далі. 

Ви можете зіткнутися з проблемами, коли ваше ядро ​​застряє на надто довгий час, ваш ноутбук працює дуже повільно та не реагує, або ваше ядро ​​втрачає з’єднання. Якщо це сталося, спробуйте виконати такі дії:
1. У верхній частині екрана натисніть **Ядро**, а потім **Перервати ядро**.
2. Якщо це не допомогло, натисніть **Ядро**, а потім **Перезапустити ядро**. Якщо ви це зробите, вам доведеться запускати клітинки коду від початку блокнота до місця, де ви призупинили роботу.

## 2.1. Арифметика
Рядок у наступній комірці віднімає числа.  Запустіть його.

```python
3.25 - 1.5
```

У Python вбудовано багато основних арифметичних операцій.  Розділ підручника про [Вирази] (http://www.inferentialthinking.com/chapters/03/1/Expressions.html) описує всі арифметичні оператори, які використовуються в курсі.  Часто вживаний оператор, який відрізняється від типової математичної нотації, — це "**", який підносить одне число до степеня іншого. Отже, `2**3` означає $2^3$ і має значення 8. 

Порядок операцій такий самий, як і в початковій школі, і в Python також є дужки.  Наприклад, порівняйте результати комірок нижче. У другій комірці використовуються круглі дужки для привітання з новим роком!


```python
9+6*5-6*3**2*2**3/4*7
```

```python
9+(6*5-(6*3))**2*((2**3)/4*7)
```

У стандартній математичній нотації перший вираз:

$$9 + 6 \times 5 - 6 \times 3^2 \times \frac{2^3}{4} \times 7,$$

тоді як другий вираз є

$$9 + (6 \times 5 - (6 \times 3))^2 \times \left(\frac{\left(2^3\right)}{4} \times 7\right).$$

**Запитання 2.1.1.** Напишіть у наступній комірці вираз Python, який дорівнює $\displaystyle 5 \times \left(3 \frac{10}{11}\right) - 50 \frac{1}{3} + 2^{0,5 \times 22} - \frac{7}{33} + 9$.  Це п’ять помножити на три і десять одинадцятих, мінус п’ятдесят і третя, плюс два у ступені половини двадцяти двох, мінус сім тридцять третіх плюс дев’ять. 

Замініть крапки (`...`) своїм виразом.  Намагайтеся використовувати дужки лише тоді, коли це необхідно.

**Примітка.** Під "$\displaystyle 3 \frac{10}{11}$" ми маємо на увазі $\displaystyle 3+\frac{10}{11}$, а не $\displaystyle 3 \times \frac{10}{11}$.

*Підказка:* Правильний результат має починатися зі знайомого числа.

№ 3. Імена
У природній мові ми маємо термінологію, яка дозволяє швидко посилатися на дуже складні поняття.  Ми не говоримо: «Це великий ссавець із коричневим хутром і гострими зубами!»  Замість цього ми просто кажемо: «Ведмідь!»

У Python ми робимо це за допомогою *операторів присвоєння*. Оператор присвоєння має ім’я ліворуч від знака `=` та вираз, який потрібно обчислити, праворуч.

(«Ведмідь!» = «Великий ссавець із коричневим хутром і гострими зубами!»)

```python
ten = 3 * 2 + 4
```

Коли ви запускаєте цю клітинку, Python спочатку обчислює значення виразу в правій частині, `3 * 2 + 4`, яке є числом 10. Потім він присвоює це значення імені `ten`.  На цьому етапі код у комірці закінчується.

Після запуску цієї комірки значення 10 прив’язується до імені `ten`:

```python
ten
```

Твердження `ten = 3 * 2 + 4` не означає, що `ten` уже дорівнює `3 * 2 + 4`, як ми могли б очікувати за аналогією з математичним записом.  Натомість цей рядок коду змінює те, що означає `ten`; тепер воно відноситься до значення 10, тоді як раніше воно взагалі нічого не означало.

# 5. Розуміння вкладених виразів (Nested Expressions)
Виклики функцій і арифметичні вирази можуть самі містити вирази.  Ви бачили приклад в останньому питанні:

    abs(42-34)

має 2 числові вирази у виразі віднімання у виразі виклику функції.  І ви, ймовірно, написали щось на зразок `abs(7-10)`, щоб обчислити `num_avenues_away`.

Вкладені вирази можуть перетворитися на складний на вигляд код. Однак спосіб розкладання складних виразів є дуже регулярним.

Припустімо, що нас цікавить дуже незвичайна довжина котів.  Ми скажемо, що довжина є незвичайною в тій мірі, наскільки вона далеко знаходиться на числовій прямій від середньої довжини кота. Приблизна середня довжина кота (серед усіх котів на Землі сьогодні) становить **18,2** дюйма.

Отже, якщо кіт Равіолі має довжину 21,7 дюйма, то його довжина відхиляється на $|21,7 - 18,2|$, або $3,5$, дюймів від середнього.  Ось зображення цього:

І ось як ми запишемо цей вираз в одному рядку коду Python:

```python
abs(21.7 - 18.2)
```

Що тут відбувається?  `abs` приймає лише один аргумент, тому всі елементи в дужках є частиною цього *єдиного аргументу*.  Зокрема, аргументом є значення виразу `21,7 - 18,2`.  Значення цього виразу дорівнює 3,5.  Це значення є аргументом для "abs".  Його абсолютне значення дорівнює 3,5, тому 3,5 є значенням повного виразу abs(21,7 - 18,2).

Малюнок, що спрощує вираз у кілька кроків:

1. `abs(21,7 - 18,2)`
2. `abs(3.5)`
3. «3,5».

Насправді це те, що Python робить для обчислення значення виразу.

## 5.1. Більше вкладень
Тепер скажімо, що ми хочемо визначити з двох довжин котів більш незвичну.  Ми використаємо функцію `max`, яка (знову ж таки) приймає два числа як аргументи та повертає більший із двох аргументів.  Поєднавши це з функцією `abs`, ми можемо обчислити більшу відстань від середнього серед двох довжин:

```python
# Просто прочитайте та запустіть цю клітинку.

ravioli_length_in = 21.7
genghis_length_in = 16.7
average_cat_length = 18.2

# Більша відстань від середньої довжини кота серед двох довжин:
larger_distance_in = max(abs(ravioli_length_in - average_cat_length), abs(genghis_length_in - average_cat_length))

# Виводимо наші результати в зручному для читання форматі:
print("The larger distance from the average length among these two cats is", larger_distance_in, "inches.")
```

Рядок, у якому обчислюється `lager_distance_in`, виглядає складним, але ми можемо розбити його на простіші компоненти, як робили раніше.

Основний рецепт полягає в багаторазовому спрощенні невеликих частин виразу:
* **Базові вирази:** Почніть із виразів, значення яких ми знаємо, як-от імена чи числа.
    - Приклади: `genghis_length_in` або `16.7`.
* **Знайдіть наступну найпростішу групу виразів:** знайдіть основні вирази, які безпосередньо пов’язані один з одним. Це може бути арифметично або як аргументи виклику функції. 
    - Приклад: `genghis_length_in - average_cat_length`.
* **Оцінити цю групу:** Оцінити арифметичний вираз або виклик функції. Використовуйте обчислене значення для заміни групи виразів.  
    - Приклад: `genghis_length_in - average_cat_length` стає `-1,3`.

```python
import pandas as pd

# 1. Запускаємо перевірку та отримуємо об'єкт результатів
results_obj = grader.check_all()

# 2. ЗБЕРЕЖЕННЯ ДЕТАЛЬНОГО ТЕКСТОВОГО ЗВІТУ (ЛОГУ ПОМИЛОК)
# Метод str() автоматично витягує весь той детальний текст із "Test case failed", "Expected", "Got" та NameError
detailed_log = str(results_obj)

with open("grading_report.txt", "w", encoding="utf-8") as text_file:
    text_file.write(detailed_log)

# 3. ЗБЕРЕЖЕННЯ БАЛІВ У ТАБЛИЦЮ CSV
# Витягуємо загальний бал, максимум, а також бали за кожне окреме питання
grades_data = {
    "Question": [],
    "Score": [],
    "Possible": []
}

# Заповнюємо бали за кожне питання окремо (наприклад, q3_1_2, q3_3_1)
for q_name, q_res in results_obj.tests.items():
    grades_data["Question"].append(q_name)
    grades_data["Score"].append(q_res.score)
    grades_data["Possible"].append(q_res.possible)

# Додаємо фінальний рядок із загальним підсумком
grades_data["Question"].append("TOTAL")
grades_data["Score"].append(results_obj.total)
grades_data["Possible"].append(results_obj.possible)

# Експортуємо в CSV
df = pd.DataFrame(grades_data)
df.to_csv("student_grades.csv", index=False, encoding="utf-8-sig")

print("Успішно згенеровано два файли:")
print("1. 'student_grades.csv' — таблиця з балами.")
print("2. 'grading_report.txt' — детальний текстовий лог усіх помилок.")
```
